```python
all_data=[]

for file in csv_files:
    path=os.path.join(folder, file)
    df = pd.read_csv(path)
    all_data.append(df)
```

```python
bike_trips = pd.concat(all_data, ignore_index=True)
```

```python
bike_trips["started_at"] = pd.to_datetime(bike_trips["started_at"])
bike_trips["ended_at"] = pd.to_datetime(bike_trips["ended_at"])
```

```python
bike_trips.info()
```

```python
bike_trips["ride_length"] = round((
    bike_trips["ended_at"] - bike_trips["started_at"]
).dt.total_seconds() / 60,2)
```

```python
bike_trips["day_of_week"] = bike_trips["started_at"].dt.day_name()
```

```python
bike_trips["month"] = bike_trips["started_at"].dt.month_name()
```

```python
bike_trips["hour"] = bike_trips["started_at"].dt.hour

bike_trips["day_type"] = bike_trips["day_of_week"].isin(
    ["Saturday", "Sunday"]
).map({True: "Weekend", False: "Weekday"})
```

```python
bike_trips[
    ["started_at", "ride_length", "day_of_week", "month", "hour", "day_type"]
].head()
```

```python
bike_trips.groupby("member_casual")["ride_length"].agg(["count", "mean", "median", "max"])
```

```python
bike_trips_clean = bike_trips[
    (bike_trips["ride_length"] >= 0) &
    (bike_trips["ride_length"] <= 1440)
].copy()
```

```python
bike_trips_clean.shape
```

```python
bike_trips_clean.groupby("member_casual")["ride_length"].agg(
    ["count", "mean", "median", "max"]
)
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Fall"

bike_trips_clean["season"] = bike_trips_clean["started_at"].dt.month.apply(get_season)
```

```python
bike_trips_clean["season"]
```

```python
season_analysis = bike_trips_clean.groupby(
    ["member_casual", "season"]
).size().reset_index(name="rides")

season_analysis
```

```python
hour_analysis = bike_trips_clean.groupby(
    ["member_casual", "hour"]
).size().reset_index(name="rides")

hour_analysis.sort_values(
    ["member_casual", "rides"],
    ascending=[True, False]
)
```

```python
bike_trips_clean[
    bike_trips_clean["member_casual"] == "casual"
]["start_station_name"].value_counts().head(10)
```

```python
bike_type_analysis = (
    bike_trips_clean
    .groupby(["member_casual", "rideable_type"])
    .size()
    .reset_index(name="rides")
)

bike_type_analysis
```
